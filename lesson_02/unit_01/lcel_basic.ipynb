```python
# pip install python-dotenv openAI langchain langchain_openai langchain-community

from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from dotenv import load_dotenv

load_dotenv("../../.env")

model = ChatOpenAI(model="gpt-4o-mini")

messages = [
    SystemMessage(content="너는 물리학자야. 과학자답게 답변해줘"),
    HumanMessage(content="저기 무지개가 보인다!"),
]

model.invoke(messages)
```

```python
print(model.invoke(messages).content)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

result = model.invoke(messages)
parser.invoke(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate

# 프롬프트 템플릿 정의: 주어진 주제에 대한 설명 요청

prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 시인이야. 시인답게 답변해줘"),
    ("human", "{user_input}"),
])

# 프롬프트, 모델, 출력 파서를 체인으로 연결
chain = prompt | model | parser

# 응답을 호출
print(chain.invoke({"user_input":"저기 무지개가 보인다!"}))
```

```python
# <스트림(Stream) 출력>

# 응답을 토큰 단위로 스트리밍하여 출력
for token in chain.stream({"user_input":"저기 무지개가 보인다!"}):
    # 스트리밍된 내용을 출력, 각 내용을 붙여서 출력하며, 버퍼를 즉시 플러시하여 실시간으로 보여줌
    print(token, end="", flush=True)
```

- ChatPromptTemplate.from_template()은 단일 문자열을 받아 기본적으로 사용자(human) 메시지로 변환하는 간편 메서드
- ChatPromptTemplate.from_messages()는 메시지 리스트를 받아 system, human, ai 등 역할(role)을 세밀하게 구성하는 메서드

```python
prompt = ChatPromptTemplate.from_template("주제 {topic}에 대해 짧게 설명을 해주세요")

# 프롬프트, 모델, 출력 파서를 체인으로 연결
from_template_chain = prompt | model | parser

# 응답을 호출
from_template_chain.invoke({"topic":"IPO"})
```

- 랭체인에서는 파이프라인을 사용한 체인을 구성하는 객체를 Runnable객체라고 함
- 이미 구성된 체인을 다른 Runnable객체와 결합할 수 있음

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


analysis_prompt = ChatPromptTemplate.from_template("이 대답을 영어로 번역해 주세요: {korean_answer}")

# 이전에 정의된 체인(chain)을 사용하여 대답을 생성하고, 그 대답을 영어로 번역하도록 프롬프트에 전달한 후, 모델을 통해 결과를 생성하여 최종적으로 문자열로 파싱하는 체인을 구성
composed_chain = {"korean_answer": from_template_chain} | analysis_prompt | model | StrOutputParser()
# "더블딥"이라는 주제로 대답을 생성하고, 체인 실행
composed_chain.invoke({"topic": "IPO"})
```

<람다 함수를 사용하여 Runnable객체 생성할 수 있다.>

```python
model = ChatOpenAI(model="gpt-4o-mini")
prompt = ChatPromptTemplate.from_template("{subject}에 대해 간단히 설명을 해주세요.")
chain = prompt | model | StrOutputParser()
translate_prompt = ChatPromptTemplate.from_template("다음 문장을 영어로 번역해 주세요: {sentence}")

# 람다 함수를 사용한 체인 구성
composed_chain = (
    # 이전에 정의된 체인(chain)을 사용하여 입력된 데이터를 받아옵니다.
    chain
    # 입력된 데이터를 "answer" 키로 변환하는 람다 함수를 적용합니다.
    | (lambda input: {"sentence": input})
    # "sentence" 키를 가진 데이터를 영어로 번역하도록 프롬프트에 전달합니다.
    | translate_prompt
    # 프롬프트에서 생성된 요청을 모델에 전달하여 결과를 생성합니다.
    | model
    # 모델에서 반환된 결과를 문자열로 파싱합니다.
    | StrOutputParser()
)
# "더블딥"라는 주제로 대답을 생성하고, 그 대답을 영어로 번역합니다.
composed_chain.invoke({"subject": "콩쥐팥쥐"})
```

< pipe()를 통해 체인 구성하기 >

```python
composed_chain_with_pipe = (
  chain
  .pipe(lambda input: {"sentence": input})
  .pipe(translate_prompt)
  .pipe(model)
  .pipe(StrOutputParser())
)

composed_chain_with_pipe.invoke({"subject": "흥부와놀부"})
```

```python
composed_chain_with_pipe = chain.pipe(lambda input:{"sentence":input}, translate_prompt, model, StrOutputParser())
composed_chain_with_pipe.invoke({"subject": "심청전"})
```

```python
from langchain_core.runnables import RunnablePassthrough


prompt = ChatPromptTemplate.from_template("{subject}에 대해 간단히 설명을 해주세요.")

output_parser = StrOutputParser()
model = ChatOpenAI(model="gpt-4o-mini")

chain = (
    {"subject": RunnablePassthrough()}  # 입력 받은 주제를 그대로 통과시킴
    | prompt                         # 프롬프트 템플릿을 적용
    | model                          # 모델을 사용해 응답 생성
    | output_parser                  # 응답을 문자열로 파싱
)

# "더블딥" 주제로 설명 요청
chain.invoke("홍길동전")
```
