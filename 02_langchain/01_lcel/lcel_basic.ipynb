```python
# pip install python-dotenv openAI langchain langchain_openai langchain-community

from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from dotenv import load_dotenv

load_dotenv("../../.env")

model = ChatOpenAI(model="gpt-4o-mini")

messages = [
    SystemMessage(content="너는 물리학자야. 과학자답게 답변해줘"),
    HumanMessage(content="저기 무지개가 보인다!"),
]

model.invoke(messages)
```

```python
print(model.invoke(messages).content)
```

#### 모델 출력에서 AI응답만 걸러내는 파서 사용

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

result = model.invoke(messages)
parser.invoke(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate

# 프롬프트 템플릿 정의: 주어진 주제에 대한 설명 요청

prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 시인이야. 시인답게 답변해줘"),
    ("human", "{user_input}"),
])

# 프롬프트, 모델, 출력 파서를 체인으로 연결
chain = prompt | model | parser

# 응답을 호출
print(chain.invoke({"user_input":"저기 무지개가 보인다!"}))
```

```python
# <스트림(Stream) 출력>

# 응답을 토큰 단위로 스트리밍하여 출력
for token in chain.stream({"user_input":"저기 무지개가 보인다!"}):
    # 스트리밍된 내용을 출력, 각 내용을 붙여서 출력하며, 버퍼를 즉시 플러시하여 실시간으로 보여줌
    print(token, end="", flush=True)
```

- ChatPromptTemplate.from_template()은 단일 문자열을 받아 기본적으로 사용자(human) 메시지로 변환하는 간편 메서드
- ChatPromptTemplate.from_messages()는 메시지 리스트를 받아 system, human, ai 등 역할(role)을 세밀하게 구성하는 메서드

```python
prompt = ChatPromptTemplate.from_template("주제 {topic}에 대해 짧게 설명을 해주세요")

# 프롬프트, 모델, 출력 파서를 체인으로 연결
from_template_chain = prompt | model | parser

# 응답을 호출
from_template_chain.invoke({"topic":"IPO"})
```

- 랭체인에서는 파이프라인을 사용한 체인을 구성하는 객체를 Runnable객체라고 함
- 이미 구성된 체인을 다른 Runnable객체와 결합할 수 있음

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


analysis_prompt = ChatPromptTemplate.from_template("이 대답을 영어로 번역해 주세요: {korean_answer}")

# 이전에 정의된 체인(chain)을 사용하여 대답을 생성하고, 그 대답을 영어로 번역하도록 프롬프트에 전달한 후, 모델을 통해 결과를 생성하여 최종적으로 문자열로 파싱하는 체인을 구성
composed_chain = {"korean_answer": from_template_chain} | analysis_prompt | model | StrOutputParser()
# "IPO"이라는 주제로 대답을 생성하고, 체인 실행
composed_chain.invoke({"topic": "IPO"})
```

<람다 함수를 사용하여 Runnable객체 생성할 수 있다.>

```python
model = ChatOpenAI(model="gpt-4o-mini")
prompt = ChatPromptTemplate.from_template("{subject}에 대해 간단히 설명을 해주세요.")
chain = prompt | model | StrOutputParser()
translate_prompt = ChatPromptTemplate.from_template("다음 문장을 영어로 번역해 주세요: {sentence}")

# 람다 함수를 사용한 체인 구성
composed_chain = (
    # 이전에 정의된 체인(chain)을 사용하여 입력된 데이터를 받아옵니다.
    chain
    # 입력된 데이터를 "answer" 키로 변환하는 람다 함수를 적용합니다.
    | (lambda input: {"sentence": input})
    # "sentence" 키를 가진 데이터를 영어로 번역하도록 프롬프트에 전달합니다.
    | translate_prompt
    # 프롬프트에서 생성된 요청을 모델에 전달하여 결과를 생성합니다.
    | model
    # 모델에서 반환된 결과를 문자열로 파싱합니다.
    | StrOutputParser()
)
# "콩쥐팥쥐"라는 주제로 대답을 생성하고, 그 대답을 영어로 번역합니다.
composed_chain.invoke({"subject": "콩쥐팥쥐"})
```

< pipe()를 통해 체인 구성하기 >

```python
composed_chain_with_pipe = (
  chain
  .pipe(lambda input: {"sentence": input})
  .pipe(translate_prompt)
  .pipe(model)
  .pipe(StrOutputParser())
)

composed_chain_with_pipe.invoke({"subject": "흥부와놀부"})
```

```python
composed_chain_with_pipe = chain.pipe(lambda input:{"sentence":input}, translate_prompt, model, StrOutputParser())
composed_chain_with_pipe.invoke({"subject": "심청전"})
```

```python
from langchain_core.runnables import RunnablePassthrough


prompt = ChatPromptTemplate.from_template("{subject}에 대해 간단히 설명을 해주세요.")

output_parser = StrOutputParser()
model = ChatOpenAI(model="gpt-4o-mini")

chain = (
    {"subject": RunnablePassthrough()}  # 입력 받은 주제를 그대로 통과시킴
    | prompt                         # 프롬프트 템플릿을 적용
    | model                          # 모델을 사용해 응답 생성
    | output_parser                  # 응답을 문자열로 파싱
)

# "홍길동전" 주제로 설명 요청
chain.invoke("홍길동전")
```

- Runnable 객체

```python
from langchain_core.runnables import RunnableLambda


def make_intro(person: str) -> str:
    """대상을 입력하면 인사말을 만드는 함수"""
    return f"{person} 안녕?"


# RunnableLambda로 감싸서 Runnable로 만들기
intro_runnable = RunnableLambda(make_intro)

# 다양한 방식으로 실행 가능
result = intro_runnable.invoke("철수")
print(result)

# 배치 처리도 자동으로 지원
results = intro_runnable.batch(["영희", "짱구", "민수"])
print(results)
```

```python
from langchain_core.runnables import RunnablePassthrough, RunnableParallel
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    "주어진 '{category}'에 포함되는 단어 3가지를 나열해주세요. 단어만 나열합니다."
)
model = ChatOpenAI(temperature=1.0, model="gpt-4o-mini")
parser = StrOutputParser()

# ① 병렬처리 체인 구성
chain = RunnableParallel(
    {
        "category": RunnablePassthrough(),
        "items": prompt | model | parser,
    }
)

result = chain.invoke({"category": "과일"})
print(result['category'])
print(result['items'])
```

```python
prompt = ChatPromptTemplate.from_template(
    "주어진 '{word}'와 유사한 단어 3가지를 나열해주세요. 단어만 나열합니다."
)
model = ChatOpenAI(temperature=1.0, model="gpt-4o-mini")
parser = StrOutputParser()

# ① 여러 분석을 동시에 수행
analysis_chain = RunnableParallel(
    synonyms=prompt | model | parser,
    word_count=RunnableLambda(lambda x: len(x["word"])),
    uppercase=RunnableLambda(lambda x: x["word"].upper()),
)

result = analysis_chain.invoke({"word": "joy"})
print(result)
```

```python
from langchain_core.runnables import RunnableBranch


# ① 입력된 텍스트가 영어인지 확인하는 함수
# 모든 문자가 ASCII 범위(128 미만)에 있으면 영어로 간주합니다.
def is_ascii(x: dict) -> bool:
    """입력 딕셔너리의 'word' 키에 해당하는 값이 영어인지 확인합니다."""
    return all(ord(char) < 128 for char in x["word"])


# ② 영어 단어에 대한 프롬프트 템플릿입니다.
english_prompt = ChatPromptTemplate.from_template(
    "Give me 3 synonyms for {word}. Only list the words."
)

# ③ 한국어 단어에 대한 프롬프트 템플릿입니다.
korean_prompt = ChatPromptTemplate.from_template(
    "주어진 '{word}'와 유사한 단어 3가지를 나열해주세요. 단어만 나열합니다."
)

# ④ 조건부 분기를 정의합니다.
# is_ascii 함수가 True를 반환하면 english_prompt를, 그렇지 않으면 korean_prompt를 사용합니다.
language_aware_chain = RunnableBranch(
    (is_ascii, english_prompt | model | parser),  # 조건이 참일 때
    korean_prompt | model | parser,  # 기본값 (조건이 거짓일 때 실행될 체인)
)

# 영어 단어 예시
english_word = {"word": "happy"}
english_result = language_aware_chain.invoke(english_word)
print(f"Synonyms for '{english_word['word']}':\n{english_result}\n")

# 한국어 단어 예시
korean_word = {"word": "행복"}
korean_result = language_aware_chain.invoke(korean_word)
print(f"'{korean_word['word']}'의 동의어:\n{korean_result}\n")
```
