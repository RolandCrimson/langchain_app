```python
from util_tools import get_current_time, get_yf_stock_history
tools = [get_current_time, get_yf_stock_history]
tool_dict = {"get_current_time": get_current_time, "get_yf_stock_history": get_yf_stock_history}

model_with_tools = model.bind_tools(tools)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage("너는 사용자의 질문에 답변을 하기 위해 tools를 사용할 수 있다."),
    HumanMessage("부산은 지금 몇시야?"),
]

response = model_with_tools.stream(messages)

# 파편화된 tool_call 청크를 하나로 합치기 
is_first = True
assembled = None
for chunk in response:    
    print("chunk type: ", type(chunk))
    
    if is_first:
        is_first = False
        assembled = chunk
    else:
        assembled += chunk
    
    print("content: ", assembled.content, "tool_call_chunk", assembled.tool_calls)

messages.append(assembled)
```

```python
assembled
```

```python
for tool_call in assembled.tool_calls:
    selected_tool = tool_dict[tool_call["name"]]
    print(tool_call["args"])
    tool_msg = selected_tool.invoke(tool_call)
    messages.append(tool_msg)

messages
```

```python
for c in model_with_tools.stream(messages):
    print(c.content, end='')
```
